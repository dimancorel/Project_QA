```python
def check_access(age, has_ticket):
    if age >= 18 and has_ticket:
        return 200
    return 403

import pytest
@pytest.mark.parametrize("age,has_ticket,expected", [
    (17,True,403),
    (18,True,200),
    (18,False,403),
    (17,False,403),
    (20,True,200)
])
def test_check_access(age, has_ticket, expected):
    assert check_access(age, has_ticket) == expected
```

```python
import requests


def test_get_post():
    response = requests.get(
        "https://jsonplaceholder.typicode.com/posts/1",
        timeout=10
    )

    assert response.status_code == 200

    data = response.json()

    assert data["id"] == 1
    assert data["userId"] == 1
    assert isinstance(data["title"], str) and len(data["title"]) > 0
```

```python
import requests
data = requests.get("https://jsonplaceholder.typicode.com/posts/2", timeout = 10)

def test_get_second_post():
    assert data.status_code == 200
    data_python = data.json()
    assert data_python["id"] == 2
    assert data_python["userId"] == 1
    assert isinstance(data_python["body"], str) and len(data_python["body"]) > 0
```

```python
import pytest

@pytest.fixture(scope="session")
def api_config():
    return {
        "base_url": "https://jsonplaceholder.typicode.com",
        "timeout": 10
    }

import requests

def test_existing_post(api_config):
    base_url = api_config["base_url"]
    response = requests.get(f"{base_url}/posts/1", timeout=api_config["timeout"])
    assert response.status_code == 200

def test_missing_post(api_config):
    base_url = api_config["base_url"]
    response = requests.get(f"{base_url}/posts/101", timeout=api_config["timeout"])
    assert response.status_code == 404
```
