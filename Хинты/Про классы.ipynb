```python
Для тех, кто не все сразу понял (как я). Код:

class Counter:
  x = 0 - пусть у класса сразу будет один параметр
  print('Counter created!') - в определение класса добавим print

c1 = Counter()
c2 = Counter()  - создадим два экземпляра класса
c2.x += 1 - одному из двух экземпляров параметр поменяем
Counter.y = 789 - а вот здесь добавим параметр уже не экземпляру, а всему классу

print(c1.x)
print(c2.x)
print(c2.y)
print(c1.y)

Выведет:

Counter created! - один раз. Когда создался сам класс. А не каждый из 2 экземпляров.
0 - у экземпляра c1 икс остался нулевым
1 - у экземпляра c2 икс поменялся
789 - когда приписываешь (или меняешь) параметр не экземпляру, а всему классу
789   он появляется (или меняется) у всех экземпляров сразу

исполнение конструктора класса начинается при создании первого экземпляра класса
```

```python
class A:
    print('исполняю тело класса а')

def a():
    print('исполняю функцию а')
при запуске скрипта будет выведено: исполняю тело класса а

Тело функции исполняется в момент ее вызова. Когда вы просто создаете функцию, описывая (определяя ее) она не исполняется. А вот класс - исполняется и соответственно, создается пространство имен под него.
```

```python
Вот правильный ответ:
"Исполнение конструктора класса начинается в момент создания нового объекта этого класса. Конструктор является специальным методом, который обязательно вызывается при создании объекта. Он отвечает за инициализацию переменных и установку начальных значений для нового экземпляра класса. Конструктор вызывается автоматически при создании объекта и возвращает готовый к использованию экземпляр класса."
```
