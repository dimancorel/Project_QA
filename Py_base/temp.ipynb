```python
import requests

adress = 'https://stepik.org/media/attachments/course67/3.6.3/'

r = requests.get(adress + '699991.txt')
filename = r.text.strip()

print('первый файл:', repr(filename))

while r.text.split()[0] != "We":
    print('скачиваем:', filename)

    r = requests.get(adress + filename)

    print('status:', r.status_code)
    print('содержимое:', repr(r.text[:100]))

    filename = r.text.strip()

print('ПОСЛЕДНИЙ ФАЙЛ:')
print(r.text)
```
