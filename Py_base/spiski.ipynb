```python
numbers = [int(x) for x in input().split()]
numbers.sort()
if len(numbers) != 1:
    n = len(numbers)
    prev = ''
    for i in range(n):
        left = numbers[i - 1]          # левый сосед (для 0-го это последний)
        right = numbers[(i + 1) % n]   # правый сосед (по кругу)
        if left == numbers[i] or right == numbers[i]:
            if prev != numbers[i]:
                print(numbers[i], end=' ')
                prev = numbers[i]
    print()
```

```python
n = int(input())
count = 0
for i in range(1, n+1):
    b = [i]*i
    if count == n:
        break
    for x in b:
        print(x, end=' ')
        count +=1
        if count == n:
            break
```

```python
matrix = []

while True:
    row = input()
    if row == 'end':
        break
    matrix.append([int(x) for x in row.split()])

n = len(matrix)
m = len(matrix[0])

for i in range(n):
    for j in range(m):
        result = 0

        left = matrix[i][j-1]
        if j == m - 1:
            right = matrix[i][0]
        else:
            right = matrix[i][j + 1]

        top = matrix[i-1][j]

        if i == n - 1:
            down = matrix[0][j]
        else:
            down = matrix[i+1][j]

        result = left + right + top + down
        print (result, end=' ')
    print (end='\n')

    #n — строк,
```
