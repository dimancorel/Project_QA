```python
Обзор всех основных методов и функций работы с множествами (set) в Python.

🌟 Множества в Python (set) — это неупорядоченная коллекция уникальных элементов.
s = {1, 2, 3}

🛠️ Создание множеств

s = set() # пустое множество
s = {1, 2, 3} # множество с элементами

📚 Основные методы множества

add(el)Добавляет элементs.add(4) → {1, 2, 3, 4}

remove(el)Удаляет элемент, вызывает ошибку, если его нетs.remove(2)

discard(el)Удаляет элемент, не вызывает ошибкуs.discard(10)

pop()Удаляет случайный элемент и возвращает егоs.pop()

clear()Очищает множествоs.clear()


🔁 Операции над множествами

1. Объединение (union)

a | b # синтаксис

a.union(b) # метод

📌 Возвращает множество, содержащее элементы из обоих.

2. Пересечение (intersection)

a & b

a.intersection(b)

📌 Только общие элементы.

3. Разность (difference)

a - b

a.difference(b)

📌 Только элементы из a, которых нет в b.

4. Симметрическая разность (symmetric_difference)

a ^ b

a.symmetric_difference(b)

📌 Все элементы, кроме общих.



⚖️ Сравнение множеств

issubset(other)Является подмножеством?a.issubset(b)

issuperset(other)Является надмножеством?a.issuperset(b)

isdisjoint(other)Нет общих элементов?a.isdisjoint(b)



📌 Функции для работы с множествами

len(s)Количество элементов

in / not inПроверка наличия элемента

set(iterable)Преобразование в множество

frozenset()Неизменяемое множество
```
