```python
# Считываем размеры поля и количество бомб
row, col, bombs = (int(i) for i in input().split())

# Создаём игровое поле с начальными значениями равными 0
a = [[0 for j in range(col)] for i in range(row)]

# Расставляем бомбы на игровом поле
for i in range(bombs):
    # Считываем координаты бомбы (введенные координаты сдвигаем на 1 влево, так как индексация начинается с 0)
    r, c = (int(i) - 1 for i in input().split())
    a[r][c] = -1

# Проходим по всему игровому полю
for i in range(row):
    for j in range(col):
        # Если в данной ячейке нет бомбы
        if a[i][j] == 0:
            # Проверяем все соседние ячейки
            for di in range(-1, 2):
                for dj in range(-1, 2):
                    ai = i + di
                    aj = j + dj
                    # Проверяем, что соседняя ячейка находится в пределах игрового поля и в ней есть бомба
                    if 0 <= ai < row and 0 <= aj < col and a[ai][aj] == -1:
                        # Увеличиваем значение текущей ячейки на 1 (счётчик бомб вокруг)
                        a[i][j] += 1

# Выводим итоговое игровое поле
for i in range(row):
    for j in range(col):
        # Если в ячейке бомба, выводим *
        if a[i][j] == -1:
            print('*', end='')
        # Если вокруг ячейки нет бомб, выводим .
        elif a[i][j] == 0:
            print('.', end='')
        # Иначе выводим количество бомб вокруг
        else:
            print(a[i][j], end='')
    # Переход на новую строку после вывода строки игрового поля
    print()
```
