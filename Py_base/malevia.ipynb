```python
from math import sqrt
forma = str(input())
if forma == 'треугольник':
    a, b, c = int(input()), int(input()), int(input())
    p = float(a+b+c)/2
    S=sqrt(p*(p-a)*(p-b)*(p-c))
    print(S)
elif forma == 'круг':
    r, pi = int(input()), 3.14
    S=pi * r ** 2
    print(S)
elif forma == 'прямоугольник':
    a, b = int(input()), int(input())
    S=a * b
    print(S)
else: print('введите тип фигуры')
```

```python
PI = 3.14

def calculate_area_circle(r: float) -> float:
    return PI * r ** 2

def calculate_area_rectangle(a: float, b: float) -> float:
    return a * b

def calculate_area_triangle(a: float, b: float, c: float) -> float:
    p = (a + b + c) / 2
    return (p * (p - a) * (p - b) * (p - c)) ** 0.5

def main():
    geometric_figure: str = input()

    if geometric_figure == "круг":
        radius = float(input())
        print(calculate_area_circle(radius))

    elif geometric_figure == "прямоугольник":
        length = float(input())
        width = float(input())
        print(calculate_area_rectangle(length, width))

    elif geometric_figure == "треугольник":
        side_1 = float(input())
        side_2 = float(input())
        side_3 = float(input())
        print(calculate_area_triangle(side_1, side_2, side_3))

    else:
        print("Неизвестная фигура")

if __name__ == "__main__":
    main()
```

```python
def triangle(a, b, c):
    p = (a + b + c) / 2
    return (p * ((p - a) * (p - b) * (p - c))) ** 0.5

form = input()
if form == 'прямоугольник':
    print(float(input()) * float(input()))
elif form == 'круг':
    print(3.14 * float(input()) ** 2)
elif form == 'треугольник':
    print(triangle(float(input()), float(input()), float(input())))
```
