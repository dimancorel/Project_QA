```python
def update_dictionary(d, key, value):
    if key in d:
        d[key].append(value)
    elif 2*key in d:
        d[2*key].append(value)
    else:
        d.update({(2*key): [value]})
```

```python
stroka = input().lower().split()
d={}
for i in stroka:
    if i not in d:
        d[i]=1
    elif i in d:
        d[i]+=1
for key, value in d.items():
    print (key, value, end='\n')
```

```python
import collections

with open('dataset_3363_3.txt', 'r') as f:
    vhod = f.read().lower().split()

c = collections.Counter(vhod)

old_key = ''
old_value = int()
for key, value in c.items():
    if value == old_value:
        if key < old_key:
            old_key = key
            old_value = value
        else:
            continue
    if value > old_value:
        old_key = key
        old_value = value
print (old_key, old_value, end='\n')
```
