```python
#Берём строку s из инпута, объявляем пустую строку ns и временную строку t, которая равна первому символу строки s. Идём циклом по срезу из строки s без первого символа. Проверяем, является ли текущий символ равным t, если да, то добавляем его к t, иначе - прибавляем первый символ t (на самом деле можно любой, там все одинаковые) с длиной строки этой же t. И t делаем равной текущему символу. Таким образом собираем новую строку с количеством символов. В конце печатаем ns, не забыв прибавить также последнюю временную строку t и ее длину, потому что итерации закончились.

s, ns = input(), ''
t = s[0]
for i in s[1:]:
  if i == t[-1]:
      t += i
  else:
      ns += t[0] + str(len(t))
      t = i
print(ns + t[0] + str(len(t)))
```
