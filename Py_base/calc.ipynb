```python
a,b,c = float(input('')),float(input('')),str(input(''))
if (c == 'mod' or c == 'div' or c == '/') and b==0: print('Деление на 0!')
elif c == 'mod': print(a%b)
elif c == 'div': print(a//b)
elif c == 'pow': print(a**b)
elif c == '+': print(a+b)
elif c == '-': print(a-b)
elif c == '/': print(a/b)
elif c == '*': print(a*b)
```

```python
operations = {
      "+": lambda x, y: x + y,
      "-": lambda x, y: x - y,
      "/": lambda x, y: x / y,
      "*": lambda x, y: x * y,
      "mod": lambda x, y: x % y,
      "pow": lambda x, y: x ** y,
      "div": lambda x, y: x // y
}

x, y = float(input()), float(input())
operation = input()

if operation in ["mod", "div", "/"] and y == 0:
    print("Деление на 0!")
else:
    print(operations[operation](x, y))
```
