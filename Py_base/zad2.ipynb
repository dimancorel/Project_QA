```python
A = int(input())
if A%400 == 0: print ('Високосный')
elif A%100 == 0: print ('Обычный')
elif A%4 == 0: print ('Високосный')
else: print ('Обычный')
```

```python
year: int = int(input())
is_leap_year: bool = (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

print("Високосный" if is_leap_year else "Обычный")
```
