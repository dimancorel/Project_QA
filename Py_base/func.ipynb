```python
lst = [1, 2, 3, 4, 5, 6]


def modify_list(l):
    for x in l:
        if x % 2 != 0:
            #l.remove(x)
            #x = 0
        elif x % 2 == 0:
            c = x // 2
            l.insert(c, x)
            l.remove(x)


print(modify_list(lst))  # None
print(lst)  # [1, 2, 3]
modify_list(lst)
print(lst)
```

```python
lst = [1, 2, 3, 4, 5, 6]


def modify_list(l):
    new_l = []
    for x in l:
        if x % 2 == 0:
            new_l.append(x // 2)
    l[:] = new_l

print(modify_list(lst))  # None
print(lst)  # [1, 2, 3]
modify_list(lst)
print(lst)
```
