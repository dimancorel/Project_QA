```python
n = int(input())
game_list = []
game_dict = {}
for i in range(n):
    game_list = input().split(';')
    #keys
    #[0] - first_team,[1] - first_team_goal,[2] - second_team,[3] - second_team_goal
    #values
    #[0] - Всего_игр,[1] - Побед,[2] - Ничьих,[3] - Поражений, [4] - Всего_очков
    if game_list[0] not in game_dict:
        game_dict [game_list[0]] = [0,0,0,0,0]
    if game_list[2] not in game_dict:
        game_dict [game_list[2]] = [0,0,0,0,0]
    game_dict[game_list[0]][0] += 1
    game_dict[game_list[2]][0] += 1
    if int(game_list[1]) > int(game_list[3]):
        game_dict[game_list[0]][1] += 1
        game_dict[game_list[0]][4] += 3
        #другой команде
        game_dict[game_list[2]][3] += 1
        game_dict[game_list[2]][4] += 0

    elif int(game_list[1]) < int(game_list[3]):
        game_dict[game_list[0]][3] += 1
        game_dict[game_list[0]][4] += 0
        #другой команде
        game_dict[game_list[2]][1] += 1
        game_dict[game_list[2]][4] += 3

    elif int(game_list[1]) == int(game_list[3]):
        game_dict[game_list[0]][2] += 1
        game_dict[game_list[0]][4] += 1
        #другой команде
        game_dict[game_list[2]][2] += 1
        game_dict[game_list[2]][4] += 1
for key, values in game_dict.items():
    print(key+':', end='')
    for x in values:
        print(x, end=' ')
    print()
```
